# Employee Salary Analysis

**Veda Technology – AI & ML Track – Task 3**

This project analyzes manually created employee salary data using Python. It calculates the total, average, highest, and lowest salary and compares the results with the original salary values.

```python
# Employee salary data (manually created)
employees = [
    {"Name": "Aarav", "Salary": 45000},
    {"Name": "Diya", "Salary": 52000},
    {"Name": "Rohan", "Salary": 38000},
    {"Name": "Isha", "Salary": 60000},
    {"Name": "Vikram", "Salary": 48000},
]

salaries = [employee["Salary"] for employee in employees]
print("Employee salary data:")
for employee in employees:
    print(f'{employee["Name"]}: ₹{employee["Salary"]:,}')
```

```python
# Calculate salary statistics using built-in Python functions
total_salary = sum(salaries)
average_salary = total_salary / len(salaries)
highest_salary = max(salaries)
lowest_salary = min(salaries)

print(f"Total Salary   : ₹{total_salary:,}")
print(f"Average Salary : ₹{average_salary:,.2f}")
print(f"Highest Salary : ₹{highest_salary:,}")
print(f"Lowest Salary  : ₹{lowest_salary:,}")
```

## Results

| Statistic | Result |
|---|---:|
| Total Salary | ₹243,000 |
| Average Salary | ₹48,600.00 |
| Highest Salary | ₹60,000 |
| Lowest Salary | ₹38,000 |

```python
# Compare calculated values with the original data
print("Original salaries:", salaries)
print("Minimum in original data:", min(salaries))
print("Maximum in original data:", max(salaries))
print("Sum of original data:", sum(salaries))
print("Calculated average:", total_salary / len(salaries))

assert total_salary == sum(salaries)
assert highest_salary == max(salaries)
assert lowest_salary == min(salaries)
assert average_salary == sum(salaries) / len(salaries)
print("\nAll calculated values match the original salary data.")
```

## Optional Pandas Analysis

Pandas can also be used to organize and analyze the same salary data.

```python
import pandas as pd

df = pd.DataFrame(employees)
print(df)

print("\nPandas salary statistics:")
print("Total:", df["Salary"].sum())
print("Average:", df["Salary"].mean())
print("Highest:", df["Salary"].max())
print("Lowest:", df["Salary"].min())
```

## Conclusion

The analysis successfully calculates the total, average, highest, and lowest employee salary. The results were also checked against the original salary list, and the values match. Pandas provides a convenient way to perform the same analysis in a tabular format.
